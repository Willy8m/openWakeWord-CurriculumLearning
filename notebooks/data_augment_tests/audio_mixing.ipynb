```python
import torch
import torchaudio
import IPython.display as ipd
from torch_audiomentations import AddBackgroundNoise

audio, sr = torchaudio.load("00e0812ca29f447e84b1039f47a91b8c.wav")

augmenter = AddBackgroundNoise(
    p=1,
    background_paths=[r"F:\data\train\real positive\mobvoi_hotword_dataset\000af5671fdbaa3e55c5e2bd0bdf8cdd.wav"], #"4883aa545befc09e38f6050fc3b2dcfe.wav"
    min_snr_in_db=10,
    max_snr_in_db=10,
    mode="per_batch"
)
mixed_audio = augmenter(audio.unsqueeze(0), sr)
mixed_audio = mixed_audio.squeeze(0)  # Remove batch dimension

# Display the audio
ipd.display(ipd.Audio(mixed_audio.numpy(), rate=sr, autoplay=True))
```

```python
import os
import IPython.display as ipd
from openwakeword.data import augment_clips
```

```python
clip_paths = ["00e0812ca29f447e84b1039f47a91b8c.wav"]
total_length = 32000
background_clip_paths = [r"F:\data\train\real positive\mobvoi_hotword_dataset\000af5671fdbaa3e55c5e2bd0bdf8cdd.wav"]
RIR_paths = [i.path for i in os.scandir(r"F:\data\train\mit_rirs")]
augmentation_probabilities = {"SevenBandParametricEQ": 0,
                              "TanhDistortion": 0,
                              "PitchShift": 0,
                              "BandStopFilter": 0,
                              "AddColoredNoise": 0,
                              "AddBackgroundNoise": 1,
                              "Gain": 0,
                              "RIR": 1}
generator = augment_clips(clip_paths, total_length, augmentation_probabilities=augmentation_probabilities, background_clip_paths=background_clip_paths, RIR_paths=RIR_paths)
mixed_audio = generator.__next__()
ipd.display(ipd.Audio(mixed_audio, rate=sr, autoplay=True))
```

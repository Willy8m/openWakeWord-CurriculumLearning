```python
from itertools import product

hyperparams = {
    "steps": [15000, 30000, 50000, 100000],
    "max_negative_weight": [2, 10, 100, 300],
    "layer_size": [32, 64]
}

# Get all permutations
all_permutations = list(product(*hyperparams.values()))

# Convert to list of dicts
param_combinations = [dict(zip(hyperparams.keys(), values)) for values in all_permutations]

# Print results
for combo in param_combinations:
    print(combo)
```

```python
import os
import uuid
import sounddevice as sd
from scipy.io.wavfile import write

def record_audio(speaker_id="", duration=2, sample_rate=16000, folder="record"):
    """
    Records audio for a given duration and saves it as a WAV file in the specified folder.

    Args:
        duration (int): Duration of the recording in seconds (default is 2 seconds).
        sample_rate (int): Sampling rate of the audio (default is 16 kHz).
        folder (str): Folder where the recorded audio will be saved.

    Returns:
        str: Path to the saved audio file.
    """
    # Ensure the folder exists
    if not os.path.exists(folder):
        os.makedirs(folder, exist_ok=False)

    # Generate a unique file name using UUID
    file_name = f"{uuid.uuid4().hex}.wav"
    file_name = str(speaker_id) + "_" + file_name if speaker_id != "" else file_name
    file_path = os.path.join(folder, file_name)

    print(f"Recording for {duration} seconds at {sample_rate} Hz...")
    
    # Record the audio
    audio_data = sd.rec(int(duration * sample_rate), samplerate=sample_rate, channels=1, dtype='int16')
    sd.wait()  # Wait until the recording is finished

    # Save the audio file
    write(file_path, sample_rate, audio_data)

    print(f"Audio saved to {file_path}")
    return file_path

def remove_file(file_path):
    if os.path.exists(file_path):
        print(f"You are about to remove {file_path}")
        confirm = input(f"Are you sure (y/n)? ")
        os.remove(file_path)
    else:
        print(f"File \"{file_path}\" doesn't exist")
```

```python
import IPython.display as ipd

file_path = record_audio(1)
ipd.display(ipd.Audio(file_path, rate=16000, normalize=True, autoplay=False))
```

```python
# Remove wav file if not correct
remove_file(file_path)
```
